## Keister example

$$
\mu = \int_{\reals^d} \cos(\lvert \vx \rvert) \exp(-\lvert \vx \rvert^2) \, \dif \vx
$$

### A banana shaped unnormalized density
The following is an example of a nontrivial densit for which it would be difficult to use a quantile function to generate random samples

```python
def log_banana(points, a=2.0, b=60.0):
    """
    Banana-shaped log-density on [0,1]^2.

    Parameters
    ----------
    points : array_like
        - shape (d,) with d=2 → returns float
        - shape (n,2) → returns array (n,)

    Returns
    -------
    float or ndarray
        log-density values (−inf outside the unit square).
    """
    pts = np.asarray(points, float)
    single = (pts.ndim == 1)
    pts = np.atleast_2d(pts)   # ensure shape (n,2)

    x, y = pts[:, 0], pts[:, 1]
    inside = (0 <= x) & (x <= 1) & (0 <= y) & (y <= 1)

    out = np.full(pts.shape[0], -np.inf, dtype=float)
    out[inside] = -a * (1 - x[inside])**2 - b * (y[inside] - x[inside]**2)**2

    return float(out[0]) if single else out

#Plot the banana
a, b = 2., 60.
def plot_banana(a=a, b=b, n_grid=200):
    xs = np.linspace(0, 1, n_grid)
    ys = np.linspace(0, 1, n_grid)
    X, Y = np.meshgrid(xs, xs)
    pts = np.column_stack([X.ravel(), Y.ravel()])
    Z = np.exp(log_banana(pts, a=a, b=b)).reshape(X.shape)  #take the exponential of the log banana
    fig, ax = plt.subplots(figsize=(7,6))
    cf = ax.contourf(X, Y, Z, levels=30, cmap="viridis")
    cs = ax.contour(X, Y, Z, levels=10, colors="white", linewidths=0.5, alpha=0.7)
    ax.set_aspect('equal')
    fig.colorbar(cf, ax=ax, label="unnormalized density")
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_title(f"Banana density, $a=$ {a}, $b=$ {b}")
    return ax

plot_banana(a=a, b=b);
```

```python
def sampler_uniform_01(d, n, rng=None):  #the uniform sampler to 
    rng = np.random.default_rng(rng)
    return rng.random((n,d))   # uniform on [0,1]^2

#Run the acceptance-rejection sampler
#Note that the uniform density proposal always lies above the target
n = 10_000
AR_samples, AR_info = accept_reject(
    log_target_density = lambda z: log_banana(z, a=a, b=b),
    proposal_sampler = lambda n, rng: sampler_uniform_01(2,n),
    n_samples = n
)
print(AR_info)
marker_size = 5
ax = plot_banana(a=a, b=b)
ax.scatter(AR_samples[:,0], AR_samples[:,1], s=marker_size, c=colors["red"], alpha=0.4, edgecolor="none")
ax.set_title(f"Banana density, $a=$ {a}, $b=$ {b}, $n=$ {n}");
```

## Markov Chain Monte Carlo (MCMC)

The weakness in acceptance-rejection sampling is that it samples from just one one proposal distribution.  Markov chain Monte Carlo overcomes this weakness by moving from positions of low density to those of higher densitiy.  Next we introduce the Metropolis-Hastings version of MCMC and contrast it with acceptance-rejection sampling.

Suppose that
* You really want to sample $\boldsymbol{X}$ with known (unnormalized) target density $\varrho$
* You have a proposal density $\varrho_{\text{new} | \text{old}}$ to select the next point starting from where you are
  
Metropolis-Hastings sampling proceeds as follows
$$
\begin{array}{l}
\text{Given } x_0 \\
\text{For } i = 0 \text{ to } n-1 \\
\begin{array}{lrl}
& \text{Generate } & z \sim \varrho_{\text{new} | \text{old}}(\cdot | x_i) \\
&& U \sim \mathcal{U}[0,1] \\
& \text{If } &U \le \min \left(1, \frac{\varrho(z) \varrho_{\text{new} | \text{old}}( x_i | z)}{\varrho(x_i) \varrho_{\text{new} | \text{old}}(z | x_i)}  \right) \\
&& x_{i+1} = z \\
& \text{Else} \\
&& x_{i+1} = x_i
\end{array}
\end{array}
$$

If $\varrho_{\text{new} | \text{old}}( x | z)= \varrho_{\text{new} | \text{old}}(z | x)$. then the decision criterion simplifies and this is called the Metropolis algorithm.

We have this algorithm in `classlib/sampling/metropolis.py`.

```python
#Some helppers
from dataclasses import dataclass
import time
import numpy as np

@dataclass
class RunSummary:
    name: str
    proposal_sd: float | None
    accept_rate: float
    elapsed_s: float
    mean: float | None = None

def run_rw_once(log_target, x0, n_samples, sd, rng=None):
    t0 = time.perf_counter()
    samples, acc = metropolis(
        log_target_density=log_target,
        x0=np.asarray(x0, float).reshape(1, -1),
        n_samples=int(n_samples),
        proposal_sd=float(sd),
        rng=rng,
    )
    elapsed = time.perf_counter() - t0
    return samples, RunSummary("RW-MH", sd, float(acc), elapsed, float(np.mean(samples)))

def safe_latex_approx_label(prefix: str, value: float) -> str:
    return rf"{prefix} $\approx$ {value:.2f}"

def print_summaries(rows: list[RunSummary]):
    for r in rows:
        sd = f"{r.proposal_sd:g}" if r.proposal_sd is not None else "-"
        mean = f"{r.mean:.3f}" if r.mean is not None else "-"
        print(f"{r.name:6s} | sd={sd:>5s} | acceptance={100*r.accept_rate:5.1f}% | mean={mean:>7s} | {r.elapsed_s*1000:.0f} ms")
```

```python
# MCMC on banana (RW–MH)
# Uses 'a' and 'b' from earlier banana section.
n = 10_000
x0 = np.array([0.5, 0.5])
std_dev = 0.10

samples, summ = run_rw_once(log_target=lambda z: log_banana(z, a=a, b=b),x0=x0,n_samples=n,sd=std_dev)
print_summaries([summ])

marker_size = 5
ax = plot_banana(a=a, b=b)
ax.scatter(samples[:, 0], samples[:, 1], s=marker_size, c=colors["red"], alpha=0.4, edgecolor="none",
           label=safe_latex_approx_label("RW MH acc", summ.accept_rate))
ax.legend()
ax.set_title(rf"Banana density, $a={a}$, $b={b}$, $n={n}$")
plt.show()
```

```python
# --- RW–MH sweeps on banana (refactor) ---
tol_keys   = ["red", "cyan", "gray", "green", "purple"]   # Paul Tol Bright keys you already use
palette    = [colors[k] for k in tol_keys]

met_cfgs = [
    {"n": 10_000, "x0": (0.5, 0.5), "sd": 1e-1},
    {"n": 1_000,  "x0": (0.5, 0.5), "sd": 1e-1},
    {"n": 10_000, "x0": (0.5, 0.5), "sd": 5e-1},
    {"n": 10_000, "x0": (0.5, 0.5), "sd": 1e-3},
    {"n": 10_000, "x0": (0.1, 0.9), "sd": 1e-3},
]

plot_every = 10
marker_size = 10
alph = 0.35

results = []
for cfg in met_cfgs:
    rng = np.random.default_rng()  # unseeded stream for this run
    samples, summ = run_rw_once(
        log_target=lambda z: log_banana(z, a=a, b=b),
        x0=np.asarray(cfg["x0"], float),
        n_samples=cfg["n"],
        sd=cfg["sd"],
        rng=rng,            # if your metropolis() ignores rng, it's still fine
    )
    results.append((cfg, samples, summ))

ax = plot_banana(a=a, b=b)
for (cfg, samples, summ), col in zip(results, palette):
    label = rf"$x_0=({cfg['x0'][0]:.2f},{cfg['x0'][1]:.2f}),\ \sigma={cfg['sd']:g},\ n={cfg['n']:.0f},\ \mathrm{{acc}}={100*summ.accept_rate:.1f}\%$"
    ax.scatter(
        samples[::plot_every, 0],
        samples[::plot_every, 1],
        s=marker_size, c=col, alpha=alph, edgecolor="none",
        label=label
    )

ax.set_title(rf"Banana RW–MH sweeps ($a={a}$, $b={b}$)")
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), frameon=False, ncol=1, fontsize=11)
ax.figure.subplots_adjust(bottom=0.12)

# concise timing/acceptance table
print_summaries([summ for _, _, summ in results])
```

## Maximum Mean Discrepancy
How do we evaluate the quality of our sample.  It is not easy.  However, we can compute the difference between two samples.

Given a symmetric, positive definite kernel, $K: \mathcal{X} \times \mathcal{X} \to \reals$, the ***discrepancy*** between two empirical distributions is defined by  

$$
D^2\bigl(\{\vx_i\}_{i=0}^{m-1},\{\vz_i\}_{i=0}^{n-1}\bigr)
= \frac{1}{m^2} \sum_{i,j=0}^{n-1} K(\vx_i,\vx_i) - \frac 2{mn} \sum_{i,j=0}^{n-1,m-1} K(\vx_i,\vz_j) + \frac{1}{n^2} \sum_{i,j=0}^{n-1} K(\vz_i,\vz_j)
$$

## Bayesian inference versus frequentist inference
We observe data 
$$
y_0,\dots,y_{n-1} \stackrel{\text{IID}}{\sim} \mathcal{N}(\mu,\sigma^2),
$$
with _known_ variance $\sigma^2$ but _unknown_ mean $\mu$.  We want to infer the mean.

A $95\%$ confidence interval for the mean is given by $\widehat{\mu}_n \pm 1.96 \sigma/\sqrt{n}$ where $\displaystyle \widehat{\mu}_n = \frac 1n \sum_{i=0}^{n-1} y_i$

### Frequentist inference
A $95\%$ confidence interval for the mean is given by $\widehat{\mu}_n \pm 1.96 \sigma/\sqrt{n}$ where $\displaystyle \widehat{\mu}_n = \frac 1n \sum_{i=0}^{n-1} y_i$

```python
n=10 #sample size
sigma = 4. #standard deviation
mu = 0.
rng = np.random.default_rng(123)  #Fixed seed for desired behavior
y = rng.normal(mu, sigma, size=n)  #data 
rng = np.random.default_rng()
print(y)
muhat = np.mean(y)
print(rf"95% confidence interval is {muhat:.3f} ± {1.96*sigma/np.sqrt(n):0.3f}")
```

### Bayesian Inference
We place a **two–component Gaussian mixture prior** on the mean:
$$
\varrho_{\text{prior}}(\mu) \propto \exp\bigl((\mu-a)^2/(2\tau^2) \bigr) \;+\;\exp\bigl((\mu+a)^2/(2\tau^2) \bigr),
\qquad a>0,\ \tau>0
$$

The __likelihood__ for $\mu$ can be expressed in terms of the sample mean $\widehat{\mu}$:
$$
\varrho_{\boldsymbol{Y}\mid \text{mean}} (\boldsymbol{y} \mid \mu)
\;\propto\; \exp\!\left(-\,\frac{(\mu-\widehat{\mu})^2}{2\sigma^2/n}\right)
$$
  
Thus the posterior is a two–component mixture:
$$
\varrho_{\text{posterior}}(\mu\mid \boldsymbol{y})
\propto w_+\;\mathcal N\!\bigl(\mu;\ m_+,\ s^2\bigr)
+ w_-\;\mathcal N\!\bigl(\mu;\ m_-,\ s^2\bigr).
$$

Here
$$
s^2=\left(\frac{1}{\tau^2}+\frac{n}{\sigma^2}\right)^{-1}, \qquad
m_{\pm}=s^2\left(\frac{\pm a}{\tau^2}+\frac{n\,\widehat{\mu}}{\sigma^2}\right).
$$

The weights are proportional to the marginal likelihood of each prior component:
$$
\tilde w_{\pm}= \exp\left( - \frac{(\widehat{\mu} - \pm a)^2}{2 \left(\tau^2+\frac{\sigma^2}{n}\right)} \right)
\qquad
w_{\pm}=\frac{\tilde w_{\pm}}{\tilde w_{+}+\tilde w_{-}}
$$

When $\widehat{\mu} \approx 0$, with $a$ moderately large and $\tau$ small,  the posterior has **two separated modes near $\pm a$**.  

This is a good stress test for Metropolis–Hastings:  a random-walk sampler tends to get stuck in one mode,  while parallel tempering or mixture proposals can explore both.

Here is the posterior

```python
# Bimodal Bayesian posterior for μ with mixture prior
# Returns log-prior and log-likelihood
def make_log_post(y, sigma=1.0, a=6.0, tau=0.1):
    n = len(y)
    inv_sigma2 = 1.0 / (sigma**2)
    inv_tau2 = 1.0 / (tau**2)
    import math
    def log_norm(x, m, inv_s2):
        return -0.5*math.log(2*np.pi) + 0.5*math.log(inv_s2) - 0.5*inv_s2*(x-m)**2
    def log_prior_mix(mu):
        t1 = np.log(0.5) + log_norm(mu, +a, inv_tau2)
        t2 = np.log(0.5) + log_norm(mu, -a, inv_tau2)
        m  = np.maximum(t1, t2)
        return m + np.log(np.exp(t1-m) + np.exp(t2-m))  # exact log prior density
    def log_lik(mu):
        return -0.5*n*np.log(2*np.pi*sigma**2) - 0.5*inv_sigma2*np.sum((y-mu)**2)
    def log_post(mu):
        mu = np.array(mu, float).reshape(-1)[0]
        return float(log_lik(mu) + log_prior_mix(mu))

    log_post.log_prior = lambda mu: float(log_prior_mix(np.array(mu, float).reshape(-1)[0]))
    log_post.log_lik   = lambda mu: float(log_lik(np.array(mu, float).reshape(-1)[0]))
    return log_post
```

```python
#Set parameters for prior and plot the prior and posterior
a, tau = 2, 0.3

log_post = make_log_post(y, sigma=sigma, a=a, tau=tau)
"""
log_post(mu)       -> log posterior
log_post.log_prior(mu) -> log prior
log_post.log_lik(mu)   -> log likelihood
"""
mu_grid = np.linspace(-5, 5, 2000)

# Posterior density on grid
logp = np.array([log_post([m]) for m in mu_grid], dtype=float)
logp -= logp.max()
p = np.exp(logp)
p /= np.trapezoid(p, mu_grid) #numerically normalize the posterior

# Prior density from the SAME closure (no duplication), already normalized
prior = np.exp(np.array([log_post.log_prior(m) for m in mu_grid], dtype=float))

# Likelihood scaled
lik = np.exp(np.array([log_post.log_lik(m) for m in mu_grid], float))
lik /= np.trapezoid(lik, mu_grid)

plt.figure(figsize=(7,4))
plt.plot(mu_grid, p, lw=2, label="posterior")
plt.plot(mu_grid, prior, '--', lw=2, label="mixture prior")
plt.plot(mu_grid, lik, ":", lw=2, label="likelihood (scaled)")
plt.xlabel(r"$\mu$")
plt.ylabel("Density")
plt.title(r"Posterior vs. Mixture Prior")
plt.legend()
plt.show()
```

Note that the data, which is centered on the right, pulls the prior to the right.

### Parallel tempering (replica exchange) for a bimodal posterior

We run $R$ Markov chains (called *replicas*) targeting a family of **tempered posteriors**
$$
\varrho_{\text{posterior},\beta_r}(\mu \mid \boldsymbol{y}) \;\propto\; \bigl[\varrho_{\boldsymbol{Y}|\text{mean}}(\boldsymbol{y} \mid \mu)\bigr]^{\beta_r}\, \varrho_{\text{prior}}(\mu),
\qquad 1=\beta_1 \ge \beta_2 \ge \cdots \ge \beta_R > 0.
$$
- The **cold** chain has $\beta_1=1$ and targets the true posterior $\varrho_{\text{posterior}}(\mu\mid \boldsymbol{y})$.
- The **hotter** chains ($\beta<1$) flatten the landscape, making it easier to move between modes.

Each replica advances for a short block by **random-walk Metropolis** steps at its own $\beta_r$.
After every block, we attempt **swaps** of states between neighboring temperatures to pass
mode-exploring states back to the cold chain.

**Replica-exchange acceptance.**  
Suppose replicas $r$ and $s=r+1$ currently hold states $\mu_r$ and $\mu_s$ at inverse temperatures
$\beta_r$ and $\beta_s$. Propose to swap these states; accept with probability
$$
\alpha_{\text{swap}} \;=\; \min\!\left\{1,\;
\exp\!\bigl(\Delta\bigr)\right\},
\qquad
\Delta \;=\; \bigl(\beta_r-\beta_s\bigr)\,\Bigl[\log \varrho_{\text{posterior}}(\mu_s\mid \boldsymbol{y})-\log \varrho_{\text{posterior}}(\mu_r\mid \boldsymbol{y})\Bigr].
$$
This preserves the correct joint stationary distribution for all replicas and lets the
cold chain inherit long jumps discovered by hotter chains.

**Why it works here.**  
With a bimodal posterior for $\mu$ (induced by the mixture prior), a local random-walk
Metropolis sampler often **gets stuck** in one mode.  
Hot replicas (smaller $\beta$) have lower energy barriers between modes and thus cross more often;
occasional **swaps** move those cross-mode states back to the cold chain, yielding
good *mode occupancy* for the chain that targets the true posterior.



```python
# --- Parallel Tempering (compact) with RW–MH (small sd) comparison ---

# choose the small-sd RW–MH run; if 0.10 missing, use the smallest available sd
small_sd = 0.10 if 0.10 in rw_runs else min(rw_runs.keys())
rw_small = rw_runs[small_sd]["samples"]
rw_acc_small = rw_runs[small_sd]["acc"]

betas = np.geomspace(1.0, 0.3, 4)

t0 = time.perf_counter()
trace_cold, finals, stats = parallel_tempering(
    base_log_density=log_post,
    x0_list=[np.array([+4.0]), np.array([-4.0]), np.array([+4.0]), np.array([-4.0])],
    betas=betas,
    n_outer=600,
    block_len=150,
    proposal_sd=[0.30/np.sqrt(b) for b in betas]
)
elapsed = time.perf_counter() - t0

# --- metrics (use the keys you actually have) ---
cold_idx = int(np.argmax(stats["betas"]))         # cold = largest beta (typically 1.0)
swap_r   = float(stats["swap_rate"])
acc_cold = float(np.asarray(stats["acceptance_rates"]).ravel()[cold_idx])

# one-line summary
print(f"PT | swap={100*swap_r:4.1f}% | accept(cold)={100*acc_cold:4.1f}% | {elapsed*1000:6.0f} ms")

# --- plot: posterior + RW–MH(small sd) + PT(cold) ---
fig, ax = plt.subplots(figsize=(7,4))
ax.plot(mu_grid, p, lw=2, label="posterior")

# use same (reasonably fine) binning for both hists
nbins = max(60, min(160, trace_cold.size // 400))

ax.hist(
    rw_small, bins=nbins, density=True, alpha=0.35, edgecolor="none",
    label=rf"RW MH ($\sigma={small_sd:g}$, acc $\approx$ {100*rw_acc_small:.0f}\%)"
)

pt_label = rf"PT cold (swap $\approx$ {swap_r:0.2f}, acc $\approx$ {100.*acc_cold:0.0f}\%)"
ax.hist(
    trace_cold, bins=nbins, density=True, alpha=0.35, edgecolor="none",
    label=pt_label
)

ax.set_xlabel(r"$\mu$")
ax.set_ylabel(r"Posterior density, $\varrho(\mu\mid y)$")
ax.set_title("Mode hopping: RW–MH (small $\sigma$) vs Parallel Tempering (cold)")
ax.legend()
plt.tight_layout()
```

**Interpreting the figures above.**

- *Posterior curve*: We evaluate the unnormalized $\log \varrho_{\text{posterior}}(\mu\mid \boldsymbol{y})$ on a grid, stabilize by
subtracting its maximum, exponentiate, and **numerically normalize** (trapezoid rule) to get a
density with area $\approx 1$.

- *RW histogram overlay*: Histogram of the plain random-walk Metropolis samples. On this example,
it typically shows **one bump** (the chain sticks in a single mode).

- *PT cold-chain histogram overlay*: Histogram of the cold chain from parallel tempering. It
typically shows **two bumps** aligned with the posterior density’s modes, demonstrating successful
mode switching.
